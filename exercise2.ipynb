```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv("../.env")

model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    groq_api_key=os.environ.get("GROQ_API_KEY"),
)

STEP1_PROMPT = PromptTemplate(
    template="""
Convert the following research paper abstract into simple
language that a non-technical person can understand.

Research Paper Abstract:
{abstract}
""",
    input_variables=["abstract"],
)

STEP2_PROMPT = ChatPromptTemplate.from_messages([
    ("system", "You are a simple language science writer."),
    ("human", "Create a short layperson summary based on this information: {information}")
])

parser = StrOutputParser()

chain = STEP1_PROMPT | model | parser | STEP2_PROMPT | model | parser

print("Enter research paper abstract:")
abstract = input()

result = chain.invoke({
    "abstract": abstract
})

print(result)
```
